```python
import os
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
DATASET_PATH = "../data/raw/mvtec_anomaly_detection/bottle"
```

```python
print("Train folders:", os.listdir(os.path.join(DATASET_PATH, "train")))
print("Test folders:", os.listdir(os.path.join(DATASET_PATH, "test")))
```

```python
sample_image_path = os.path.join(
    DATASET_PATH,
    "test",
    "broken_large",
    os.listdir(os.path.join(DATASET_PATH, "test", "broken_large"))[0]
)

sample_image = cv2.imread(sample_image_path)
sample_image = cv2.cvtColor(sample_image, cv2.COLOR_BGR2RGB)

print("Image Shape:", sample_image.shape)
```

```python
plt.figure(figsize=(6,6))
plt.imshow(sample_image)
plt.title("Sample Defective Bottle")
plt.axis("off")
plt.show()
```

```python
good_image_path = os.path.join(
    DATASET_PATH,
    "test",
    "good",
    os.listdir(os.path.join(DATASET_PATH, "test", "good"))[0]
)

bad_image_path = os.path.join(
    DATASET_PATH,
    "test",
    "broken_large",
    os.listdir(os.path.join(DATASET_PATH, "test", "broken_large"))[0]
)

good_image = cv2.cvtColor(cv2.imread(good_image_path), cv2.COLOR_BGR2RGB)

bad_image = cv2.cvtColor(cv2.imread(bad_image_path), cv2.COLOR_BGR2RGB)
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12,6))

axes[0].imshow(good_image)
axes[0].set_title("Normal Bottle")
axes[0].axis("off")

axes[1].imshow(bad_image)
axes[1].set_title("Defective Bottle")
axes[1].axis("off")

plt.show()
```

```python
mask_path = os.path.join(
    DATASET_PATH,
    "ground_truth",
    "broken_large",
    os.listdir(os.path.join(DATASET_PATH, "ground_truth", "broken_large"))[0]
)

mask = cv2.imread(mask_path, cv2.IMREAD_GRAYSCALE)

print("Mask Shape:", mask.shape)
```

```python
plt.figure(figsize=(6,6))

plt.imshow(mask, cmap="gray")
plt.title("Ground Truth Defect Mask")
plt.axis("off")

plt.show()
```
